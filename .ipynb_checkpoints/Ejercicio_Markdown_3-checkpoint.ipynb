### Ejercicio 2: Estilos de texto
Escribe una frase que contenga las siguientes características:

1. La palabra "Markdown" en negritas.
2. La palabra "Jupyter" en cursivas.
3. La palabra "Python" en negritas y cursivas.

1. La palabra **Markdown** está en negrita
2. La palabra *Jupyter* está en cursiva
3. La palabra ***Python*** está en negrita y cursiva

**Markdown** es un lenguaje que se utiliza habitualmente en los *Jupyter* notebooks. Los *Jupyter* notebooks se suelen utilizar en lenguajes de programación como ***Python***

### Ejercicio 3: Resaltados y HTML
Crea una línea que contenga:

1. La palabra "resaltado" con un fondo amarillo usando Markdown.
2. La palabra "color" en azul usando HTML.

Para este ejercicio tendrás que buscar un poco más en internet, venga que tú puedes.

La palabra <mark>resaltado</mark> está en amarillo usando Markdown

Para usar el <mark>resaltado</mark> en Markdown tienes que utilizar `<mark>palabra<mark>`

<p>La palabra <font color="blue">color</font> está en azul usando HTML.</p>

Por otro lado, para resaltar con <font color="blue">color</font> azul usando HTML, se usa `<font color="blue">palabra</font>`
