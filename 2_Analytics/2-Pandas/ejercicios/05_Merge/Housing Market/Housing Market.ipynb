### Step 1. Import the necessary libraries

```python
import pandas as pd
import numpy as np
```

### Step 2. Create 3 differents Series, each of length 100, as follows: 
1. The first a random number from 1 to 4 
2. The second a random number from 1 to 3
3. The third a random number from 10,000 to 30,000

```python
serie1 = pd.Series(np.random.randint(1,5,size=100))
serie1
```

```python
serie2 = pd.Series(np.random.randint(1,4,size=100))
serie2
```

```python
serie3 = pd.Series(np.random.randint(10000,30001,size=100))
serie3
```

### Step 3. Let's create a DataFrame by joinning the Series by column

```python
df=pd.DataFrame([])
df=pd.concat([serie1,serie2,serie3],axis=1)
df
```

### Step 4. Change the name of the columns to bedrs, bathrs, price_sqr_meter

```python
df.columns=["bedrs","bathrs","price_sqr_meter"]
df
```

### Step 5. Create a column DataFrame with the values of the 3 Series and assign it to 'bigcolumn'

```python
bigcolumn=pd.concat([serie1,serie2,serie3],axis=0)
bigcolumn
```
