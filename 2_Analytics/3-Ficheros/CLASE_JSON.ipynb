#### Lectura con requests

````python
import json
import requests

response = requests.get(url)
todos = json.loads(response.text)
````

````python
todos
````

Vemos qué tipo de datos es esto

````python
print(type(todos))
len(todos)
````

Es un iterable, por lo que podremos recorrerlo mediante un for, y de esta manera ver qué hay dentro

````python
for i in todos:
    print(i.keys())
````

#### Obtención de datos

Primero vamos a probar a sacar todos los url.

````python
datos = {'url': []}

for i in todos:
    
    repo = i.get('repo', None)
    if repo is not None:
        nombre_url = repo.get('url', 'NO URL')
    
    datos['url'].append(nombre_url)

    
datos
````

Ahora obtendremos más info.

````python
datos = {'name': [],
        'url': [],
        'id': []
        }

for i in todos:
    repo = i.get('repo', None)
    if repo is not None:
        nombre_url = repo.get('url', 'NO URL')
        nombre_name = repo.get('name', 'NO NAME')
        nombre_id = repo.get('id', 'NO ID')
    
        datos['url'].append(nombre_url)
        datos['name'].append(nombre_name)
        datos['id'].append(nombre_id)
    
pd.DataFrame(datos).head()
````
