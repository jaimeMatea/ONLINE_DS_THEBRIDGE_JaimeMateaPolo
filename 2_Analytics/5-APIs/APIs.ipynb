#  APIs 

### Una API (Application Program Interface) es una manera que tienen los sistemas para comunicarse entre sí con software.
### REST viene de Representational State Transfer.

### Ideas de qué es una API RESTful:

#### 1. Usa métodos HTTP estándar

Son los mismos verbos que usa un navegador de internet:

**GET** → lee datos

**POST** → crea datos nuevos

**PUT / PATCH** → actualiza datos

**DELETE** → borra datos


#### 2. Todo es un “recurso”

Un recurso es algo a lo que le prestas atención:

users

products

orders

posts

**Endpoints:**

Cada recurso tiene una URL (endpoint):

/users

/products/123

/orders/456

**Posibles queries:**

**GET** /users

**POST** /users

**GET** /users/42

**DELETE** /users/42

#### 3. Es stateless (sin estado)

Cada petición (query) es independiente.

El servidor no recuerda quién eres entre petición y petición.
Cada petición debe incluir toda la información que necesite (como tokens de autenticación).

#### 4. Los datos se suelen devolver como un JSON
Un formato muy flexible para trabajar con datos.

#### 5. Un compotamiento predecible y claro

Las URL representan nombres (recursos), no acciones.

El resultado de la query (petición) tiene un estado.

El estado HTTP importa:

200 OK

201 Creado

404 No encontrado

401 No autorizado

### URI Format

* **URI** (Uniform Resource Identifier) = identificador (término paraguas)

* **URL** (Uniform Resource Locator) = identificador + ubicación

* **URN** (Uniform Resource Name) = identificador + nombre solo

* **URL** y **URN** son ambos tipos de **URI**.

````python

import requests
import time

overpass_url = "https://overpass-api.de/api/interpreter"

overpass_query = """
[out:json][timeout:25];
(
  node["amenity"="cafe"](40.43000,-3.69201,40.44447,-3.67101);
  way["amenity"="cafe"](40.43000,-3.69201,40.44447,-3.67101);
  relation["amenity"="cafe"](40.43000,-3.69201,40.44447,-3.67101);
);
out center;
"""

headers = {
    "User-Agent": "CafeFinder/1.0 (contact: alberto@thebridge.tech)"
}

try:
    response = requests.post(
        overpass_url,
        data={"data": overpass_query},
        headers=headers,
        timeout=(10, 45)
    )

    print("Status:", response.status_code)

    response.raise_for_status()
    data = response.json()

    print("Número de cafeterías:", len(data["elements"]))

except requests.exceptions.Timeout:
    print("Overpass está tardando demasiado. Inténtalo más tarde.")
except requests.exceptions.HTTPError as e:
    print("Error HTTP:", e)
    print(response.text[:500])
except requests.exceptions.RequestException as e:
    print("Error de conexión:", e)
````

````python

# Access the values in the dictionary to save the variables in a DataFrame

import pandas as pd

dict_cafes = {"amenity": [], "name":[], "lat":[], "lon":[], "street":[], "number_street":[], "opening_hours": []}
for i in data["elements"]:
    amenity = i.get('tags', None).get('amenity', None)
    name = i["tags"].get("name", "No name")
    lat = i["lat"]
    lon = i["lon"]
    street = i["tags"].get("addr:street", "No street")
    number_street = i["tags"].get("addr:housenumber", 0)
    
    opening_hours = i["tags"].get("opening_hours", "No opening hours")

    dict_cafes["amenity"].append(amenity)
    dict_cafes["name"].append(name)
    dict_cafes["lat"].append(lat)
    dict_cafes["lon"].append(lon)
    dict_cafes["street"].append(street)
    dict_cafes["number_street"].append(number_street)
    
    dict_cafes["opening_hours"].append(opening_hours)

df_cafe = pd.DataFrame(dict_cafes)
df_cafe
````

````python

# Now let's look for restaurants

import pandas as pd
import requests
import time

overpass_url = "https://overpass-api.de/api/interpreter"

headers = {
    "User-Agent": "CafeFinder/1.0 (contact: alberto@thebridge.tech)",
    "Accept": "application/json"
}

overpass_query = """
[out:json][timeout:25];
(
  node["amenity"="restaurant"](40.43000,-3.69201,40.44447,-3.67101);
  way["amenity"="restaurant"](40.43000,-3.69201,40.44447,-3.67101);
  relation["amenity"="restaurant"](40.43000,-3.69201,40.44447,-3.67101);
);
out center;
"""

# Request with exponential backoff
data = None
wait = 10

for attempt in range(5):
    try:
        response = requests.post(
            overpass_url,
            data={"data": overpass_query},
            headers=headers,
            timeout=(10, 45)
        )

        if response.status_code in (429, 502, 503, 504):
            if attempt == 4:
                response.raise_for_status()

            retry_after = response.headers.get("Retry-After")
            delay = float(retry_after) if retry_after else wait

            print(f"Server busy. Retrying in {delay:g} seconds...")
            time.sleep(delay)
            wait *= 2
            continue

        response.raise_for_status()
        data = response.json()
        break

    except requests.exceptions.Timeout:
        if attempt == 4:
            raise

        print(f"Timeout. Waiting {wait} seconds...")
        time.sleep(wait)
        wait *= 2

# Create the restaurants DataFrame
if data is not None:
    dict_restaurants = {
        "amenity": [],
        "name": [],
        "lat": [],
        "lon": [],
        "street": [],
        "number_street": [],
        "opening_hours": []
    }

    for i in data["elements"]:
        tags = i.get("tags", {})

        amenity = tags.get("amenity", None)
        name = tags.get("name", "No name")

        # Nodes have lat/lon; ways and relations use center
        if "lat" in i and "lon" in i:
            lat = i["lat"]
            lon = i["lon"]
        elif "center" in i:
            lat = i["center"]["lat"]
            lon = i["center"]["lon"]
        else:
            continue

        street = tags.get("addr:street", "No street")
        number_street = tags.get("addr:housenumber", 0)
        opening_hours = tags.get(
            "opening_hours", "No opening hours"
        )

        dict_restaurants["amenity"].append(amenity)
        dict_restaurants["name"].append(name)
        dict_restaurants["lat"].append(lat)
        dict_restaurants["lon"].append(lon)
        dict_restaurants["street"].append(street)
        dict_restaurants["number_street"].append(number_street)
        dict_restaurants["opening_hours"].append(opening_hours)

    df_restaurants = pd.DataFrame(dict_restaurants)
    print("Number of restaurants:", len(df_restaurants))
    display(df_restaurants)
````
