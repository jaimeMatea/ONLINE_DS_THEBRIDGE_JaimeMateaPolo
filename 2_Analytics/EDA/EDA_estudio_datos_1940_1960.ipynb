```python
variables = ["war_score","poverty_score","political_repression_score","nuclear_fear_score","trauma_score"]

df_evolucion = (peliculas_periodo.groupby("release_year")[variables].mean().reset_index())

df_evolucion_largo = df_evolucion.melt(id_vars="release_year",var_name="tematica",value_name="puntuacion")

fig = px.line(
    df_evolucion_largo, x="release_year", y="puntuacion", color="tematica", markers=True,
    title="Evolución de las temáticas cinematográficas (1940-1960)",
    labels={"release_year": "Año", "puntuacion": "Puntuación media (0-5)", "tematica": "Temática"}
)

fig.show()
```

```python
df_pobreza = (
    df_comparacion_peliculas_40_60
    .groupby(["country", "year"], as_index=False)
    .agg(
        poverty_score=("poverty_score", "mean"),
        infant_mortality=("infant_mortality", "mean"),
        num_peliculas=("title", "count")
    )
)

df_pobreza = df_pobreza.dropna(
    subset=["poverty_score", "infant_mortality"]
)

df_pobreza = df_pobreza.sort_values(["country", "year"])



colores = {
    "United States": "#E09F3E",
    "Japan": "#457B9D",
    "Italy": "#9B5DE5",
    "France": "#E76F51"
}

fig = make_subplots(
    rows=2,
    cols=1,
    shared_xaxes=True,
    vertical_spacing=0.12,
    subplot_titles=(
        "01 · MORTALIDAD INFANTIL",
        "02 · POBREZA EN EL CINE"
    )
)

for pais in df_pobreza["country"].unique():

    df_pais = df_pobreza[
        df_pobreza["country"] == pais
    ].sort_values("year")

    color = colores.get(pais, "#AAAAAA")

    fig.add_trace(
        go.Scatter(
            x=df_pais["year"],
            y=df_pais["infant_mortality"],
            mode="lines+markers",
            name=pais,
            legendgroup=pais,
            line=dict(color=color, width=3),
            marker=dict(size=7),
            hovertemplate=(
                f"<b>{pais}</b><br>"
                "Año: %{x}<br>"
                "Mortalidad infantil: %{y:.1f}<br>"
                "<extra></extra>"
            )
        ),
        row=1, col=1
    )

    # Representación de la pobreza
    fig.add_trace(
        go.Scatter(
            x=df_pais["year"],
            y=df_pais["poverty_score"],
            mode="lines+markers",
            name=pais,
            legendgroup=pais,
            showlegend=False,
            line=dict(color=color, width=3),
            marker=dict(
                size=7,
                line=dict(color="#F4EBDD", width=1)
            ),
            hovertemplate=(
                f"<b>{pais}</b><br>"
                "Año: %{x}<br>"
                "Pobreza cinematográfica: %{y:.2f}/5"
                "<extra></extra>"
            )
        ),
        row=2, col=1
    )

fig.update_layout(
    title=dict(
        text=(
            "INFANCIAS EN LA SOMBRA<br>"
            "<sup>Mortalidad infantil y representación de la pobreza "
            "en el cine (1939-1960)</sup>"
        ),
        x=0.5,
        font=dict(size=22, color="#F4EBDD")
    ),
    template="plotly_dark",
    paper_bgcolor="#171717",
    plot_bgcolor="#171717",
    height=850,
    width=1200,
    hovermode="x unified",
    legend=dict(
        title="Países",
        orientation="h",
        y=1.02,
        x=0.5,
        xanchor="center"
    ),
    margin=dict(t=150, b=80)
)

fig.update_yaxes(
    title_text="Fallecimientos por 1.000 nacidos vivos",
    gridcolor="#393939",
    rangemode="tozero",
    row=1, col=1
)

fig.update_yaxes(
    title_text="Presencia de la pobreza (0-5)",
    gridcolor="#393939",
    rangemode="tozero",
    row=2, col=1
)

fig.update_xaxes(
    title_text="Año",
    dtick=2,
    gridcolor="#393939",
    row=2, col=1
)

fig.show()
```

```python
df_evolucion_social = (df_comparacion_peliculas_40_60.groupby(["country", "year"], as_index=False)
    .agg({"life_expectancy": "first", "trauma_score": "mean"})
)
```

```python
import pandas as pd
import plotly.graph_objects as go
from plotly.subplots import make_subplots

# =====================================================
# 1. PREPARAR DATOS CINEMATOGRÁFICOS
# =====================================================

df_cine_anual = peliculas_periodo.copy()

df_cine_anual["country"] = (
    df_cine_anual["paises_norm"]
    .fillna("")
    .str.split(",")
)

df_cine_anual = df_cine_anual.explode("country")

df_cine_anual["country"] = df_cine_anual["country"].str.strip()

df_cine_anual = df_cine_anual[
    df_cine_anual["country"].isin(
        ["United States", "France", "Japan", "Italy"]
    )
].copy()

# Agrupar por país para calcular la media cinematográfica
df_rep_paises = (
    df_cine_anual
    .groupby("country", as_index=False)
    .agg(
        repression_mean=("political_repression_score", "mean"),
        num_peliculas=("title", "count")
    )
)


# =====================================================
# 2. PREPARAR DATOS DE MORTALIDAD
# =====================================================

df_mortalidad = df_mortalidad_wwii.copy()

df_mortalidad["total_deaths_max"] = pd.to_numeric(
    df_mortalidad["total_deaths_max"],
    errors="coerce"
)

df_mortalidad = df_mortalidad[
    ["country", "total_deaths_max"]
].drop_duplicates(subset="country")


# =====================================================
# 3. UNIR LOS DOS DATAFRAMES
# =====================================================

df_comparacion = df_rep_paises.merge(
    df_mortalidad,
    on="country",
    how="inner"
)

# Eliminar posibles valores ausentes
df_comparacion = df_comparacion.dropna(
    subset=["repression_mean", "total_deaths_max"]
)


# =====================================================
# 4. COLORES POR PAÍS
# =====================================================

colores = {
    "United States": "#E09F3E",
    "Japan": "#457B9D",
    "Italy": "#9B5DE5",
    "France": "#E76F51"
}


# =====================================================
# 5. ORDENAR LOS PAÍSES
# =====================================================

df_muertes = df_comparacion.sort_values(
    "total_deaths_max",
    ascending=True
)

df_rep = df_comparacion.sort_values(
    "repression_mean",
    ascending=True
)


# =====================================================
# 6. CREAR LOS DOS PANELES
# =====================================================

fig = make_subplots(
    rows=1,
    cols=2,
    subplot_titles=[
        "Impacto humano de la guerra",
        "Represión política en el cine"
    ],
    horizontal_spacing=0.20
)

# Panel izquierdo: muertes
fig.add_trace(
    go.Bar(
        x=df_muertes["total_deaths_max"],
        y=df_muertes["country"],
        orientation="h",
        marker_color=[
            colores[p] for p in df_muertes["country"]
        ],
        text=df_muertes["total_deaths_max"].apply(
            lambda x: f"{x:,.0f}"
        ),
        textposition="outside",
        cliponaxis=False,
        hovertemplate=(
            "<b>%{y}</b><br>"
            "Muertes estimadas: %{x:,.0f}"
            "<extra></extra>"
        ),
        showlegend=False
    ),
    row=1,
    col=1
)

# Panel derecho: represión política
fig.add_trace(
    go.Bar(
        x=df_rep["repression_mean"],
        y=df_rep["country"],
        orientation="h",
        marker_color=[
            colores[p] for p in df_rep["country"]
        ],
        text=df_rep["repression_mean"].apply(
            lambda x: f"{x:.2f}"
        ),
        textposition="outside",
        cliponaxis=False,
        customdata=df_rep[["num_peliculas"]],
        hovertemplate=(
            "<b>%{y}</b><br>"
            "Represión política media: %{x:.2f}<br>"
            "Películas: %{customdata[0]}"
            "<extra></extra>"
        ),
        showlegend=False
    ),
    row=1,
    col=2
)


# =====================================================
# 7. DISEÑO
# =====================================================

fig.update_layout(
    template="plotly_dark",
    paper_bgcolor="#171717",
    plot_bgcolor="#171717",
    font=dict(
        family="Arial",
        color="#F4EBDD"
    ),
    title=dict(
        text=(
            "DOS HUELLAS DE LA SEGUNDA GUERRA MUNDIAL"
            "<br><sup>Impacto humano y presencia de la represión "
            "política en el cine (1940-1960)</sup>"
        ),
        x=0.5,
        font=dict(size=21)
    ),
    height=550,
    margin=dict(l=100, r=100, t=130, b=70),
    showlegend=False
)

fig.update_xaxes(
    showgrid=True,
    gridcolor="#393939",
    zeroline=False
)

fig.update_yaxes(
    showgrid=False,
    autorange=True
)

fig.update_xaxes(
    title_text="Muertes estimadas",
    row=1,
    col=1
)

fig.update_xaxes(
    title_text="Puntuación media de represión (0-5)",
    row=1,
    col=2
)

fig.show()
```
