````python
import pandas as pd
import numpy as np

url = "https://exoplanetarchive.ipac.caltech.edu/TAP/sync?query=select+*+from+pscomppars&format=csv"

df = pd.read_csv(url)

print(df.shape)
````

````python
df_completo = df[["pl_name", # Nombre del exoplaneta
         "hostname", # Nombre de la estrella alrededor de la que orbita
         "pl_rade", # Radio del planeta, expresado en radios terrestres
         "pl_bmasse", # Masa del planeta, expresada en masas terrestres
         "pl_dens", # Densidad del planeta
         "pl_orbper", # Tiempo que tarda el planeta en completar una vuelta alrededor de su estrella, en días
         "pl_orbsmax", # Semieje mayor de su órbita, normalmente expresado en unidades astronómicas (UA)
         "pl_insol", # Cantidad de radiación que recibe el planeta de su estrella, comparada con la que recibe la Tierra del Sol
         "pl_eqt", # Temperatura de equilibrio estimada del planeta, expresada en Kelvin (K)
         "pl_orbeccen", # Excentricidad orbital: indica cuánto se aleja la órbita de una circunferencia perfecta
         "st_teff", # Temperatura efectiva de la estrella, expresada en Kelvin
         "st_lum", # Luminosidad de la estrella, expresada en escala logarítmica respecto al Sol
         "st_rad", # Radio de la estrella, comparado con el radio del Sol
         "st_mass", # Masa de la estrella
         "st_age", # Edad de la estrella
         "st_met", # Metalicidad de la estrella
         "st_logg", # Gravedad superficial de la estrella
         "sy_dist"]].copy() # Distancia desde nuestro sistema solar hasta el sistema planetario, expresada en pársecs (pc)
df
````

````python
print("Dimensiones:", df.shape)
````

````python
df = df.dropna(subset=["pl_rade", "pl_insol", "st_teff"]).copy()
````

````python
# Diferencia entre la temperatura de la estrella y la solar
T = df["st_teff"] - 5780

# Límite interior: Runaway Greenhouse
df["hz_inner"] = (1.107 + (1.332e-4*T) + (1.580e-8*T**2) - (8.308e-12*T**3) - (1.931e-15*T**4))

# Límite exterior: Maximum Greenhouse
df["hz_outer"] = (0.356 + (6.171e-5*T) + (1.698e-9*T**2) - (3.198e-12*T**3) - (5.575e-16*T**4))
````

````python
# Condición 1: tamaño aproximadamente terrestre
condicion_radio = df["pl_rade"].between(0.5, 1.6)

# Condición 2: radiación dentro de la zona habitable
condicion_zona = ((df["pl_insol"] >= df["hz_outer"]) & (df["pl_insol"] <= df["hz_inner"]))

# Condición 3: temperatura estelar dentro del rango del modelo
condicion_estrella = df["st_teff"].between(2600, 7200)

# Crear variable objetivo
df["habitability_class"] = (condicion_radio & condicion_zona & condicion_estrella).astype(int)
````
