## Building a regression model

The data have already been split to predictor (*data*) and response (*target*) variables. Given this information, we'll apply what we have previously learned about linear regression and apply the algorithm to the diabetes dataset.

#### In the cell below, import the `linear_model` class from `sklearn`. 

```python
from sklearn.linear_model import LinearRegression
```

#### Create a new instance of the linear regression model and assign the new instance to the variable `diabetes_model`.

```python
diabetes_model=LinearRegression()
```

#### Next, let's split the training and test data.

Define `diabetes_data_train`, `diabetes_target_train`, `diabetes_data_test`, and `diabetes_target_test`. Use the last 20 records for the test data and the rest for the training data.

```python
diabetes_data_train, diabetes_data_test, diabetes_target_train, diabetes_target_test = \
    train_test_split(diabetes.data, diabetes.target, test_size=20, random_state=42, shuffle=False)
```

```python
print("Dimensiones de los conjuntos:")
print(f"diabetes_data_train: {diabetes_data_train.shape}")
print(f"diabetes_data_test:  {diabetes_data_test.shape}")
print(f"diabetes_target_train: {diabetes_target_train.shape}")
print(f"diabetes_target_test:  {diabetes_target_test.shape}\n")
```

Perform a brief exploratory analysis to investigate the conditions necessary for linear regression. This analysis will help you understand the distribution of the target variable and its correlations with other features.

Fit the training data and target to `diabetes_model`. Print the *intercept* and *coefficients* of the model.

```python
diabetes_model.fit(diabetes_data_train, diabetes_target_train)
```

```python
a=diabetes_model.intercept_
b=diabetes_model.coef_
print(a)
print(b)
```

#### Inspecting the results

From the outputs you should have seen:

- The intercept is a float number.
- The coefficients are an array containing 10 float numbers.

This is the linear regression model fitted to your training dataset.

#### Using your fitted linear regression model, predict the *y* of `diabetes_data_test`.

```python
diabetes_target_pred = diabetes_model.predict(diabetes_data_test)
diabetes_target_pred
```

y = a + b*x1 + c*x2 + d*x3 + ....

#### Print your `diabetes_target_test` and compare with the prediction. 

```python
print(diabetes_target_test)
```

#### Is `diabetes_target_test` exactly the same as the model prediction? Use different error metrics

```python
from sklearn import metrics
from sklearn.metrics import mean_absolute_percentage_error

print('MAE:', metrics.mean_absolute_error(diabetes_target_test, diabetes_target_pred))
print('MAPE:', mean_absolute_percentage_error(diabetes_target_test, diabetes_target_pred)*100)
print('MSE:', metrics.mean_squared_error(diabetes_target_test, diabetes_target_pred))
print('RMSE:', np.sqrt(metrics.mean_squared_error(diabetes_target_test, diabetes_target_pred)))
```

```python
from sklearn.metrics import r2_score

diabetes_target_train_pred = diabetes_model.predict(diabetes_data_train)

r2_train = r2_score(diabetes_target_train, diabetes_target_train_pred)
r2_test = r2_score(diabetes_target_test, diabetes_target_pred)

print(f"R2 score train {r2_train}")
print(f"R2 score test {r2_test}")
```

```python
error=diabetes_target_test-diabetes_target_pred #Calculamos los residuos (error: valor real menos valor predicho)
sns.histplot(x=error, bins=6)
plt.xlabel("target")
plt.show()
```
